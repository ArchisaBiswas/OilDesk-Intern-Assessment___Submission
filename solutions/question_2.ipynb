Question 2: CRUD Operations in SQL Server

Objective: Basic SQL Server interactions.

```python
import sqlite3
```

```python
connection = sqlite3.connect("metals.db")
cursor = connection.cursor()
```

Creating an SQL table schema to store time-series metal prices. Including fields like Date, Metal, Price:

```python
cursor.execute("""
CREATE TABLE MetalPrices (
    Id INT AUTO_INCREMENT PRIMARY KEY,
    Date DATE NOT NULL,
    Metal VARCHAR(50) NOT NULL,
    Price DECIMAL(18,4) NOT NULL
)
""")
```
