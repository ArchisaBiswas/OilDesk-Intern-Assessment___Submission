Question 2: CRUD Operations in SQL Server

Objective: Basic SQL Server interactions.

```python
# Importing sqlite3 to interact with database, and pandas for data manipulation:
import sqlite3
import pandas as pd
```

Establishing connection and setting up through the cursor variable which will be used to interact with database in this task:

```python
connection = sqlite3.connect("metals.db")
cursor = connection.cursor()
```

Creating an SQL table schema to store time-series metal prices. Including fields like Date, Metal, Price:

A Table is created called MetalPrices. The Integrity Constraints of this Table are the Id, which is the primary key and will automatically increment so that it is not a required field for every insertion operation and the like. The Date column is of Date data type to hold the time attribute of this time-series. While the Metal and Price columns are of string to hold the Metal's name and that Metal's corresponding settlement price of that particular day respectively. Date, Metal and Price fields are Not Null, and hence, required, as these are the elements of the data-set that we want to store in this database.

```python
# Creation of Table occurs in the Database:
cursor.execute("""
CREATE TABLE MetalPrices (
    Id INTEGER PRIMARY KEY AUTOINCREMENT,
    Date DATE NOT NULL,
    Metal VARCHAR(50) NOT NULL,
    Price DECIMAL(18,4) NOT NULL
)
""")
```

To verify that the Table has been created and that it does not contain any data points at this stage:

Please Note: Retrieval is done throughout this Notebook using read_sql() in order to retrieve data from the table and then load it into a DataFrame so that it can be displayed in a tabular format in the Notebook, to enhance readability and maintainability.

```python
# Retrieval query to attempt accessing the above created Table:
df = pd.read_sql("SELECT * FROM MetalPrices", connection)
df
```

Therefore, since the result shows the Column Headers as specified with no data points inside this DataFrame at this point of time, it is hence, verified that the table Metalprices has been created and now exists in the database with no current data points in it so far.

Demonstrating basic CRUD operations:

```python
Create:
```

Making Insertions with data which is made only for the purposes of this demonstration and not sourced from anywhere specific, into the MetalPrices Table:

```python
# Making 5 data points within the Insertion query:
cursor.execute("""
INSERT INTO MetalPrices (Date, Metal, Price)
VALUES 
('2010-01-01', 'Copper', 7375),
('2010-01-01', 'Aluminum', 2200),
('2010-01-01', 'Zinc', 2500),
('2010-01-01', 'Lead', 2100),
('2010-01-01', 'Nickel', 15000);
""")
```

To verify that the above Insertions have taken place, I retrieve the values from the MetalPrices Table:

```python
df = pd.read_sql("SELECT * FROM MetalPrices", connection)
df
```

Hence, the Insertions of the values for this demonstration have taken place.

Read:

```python
# Retrieving all the data points, using "*" from the MetalPrices Table:
df = pd.read_sql("SELECT * FROM MetalPrices", connection)
df
```

Therefore, the Retrieval of all the data points has taken place, as the DataFrame df here demonstrates.

Update:

Making an Update query to modify the Price of Copper on the 1st of January, 2010 to 7400:

```python
# Updating the data point at the locations through the filter of the specified Metal and Date using the "WHERE" clause, 
# and "SET" to specify the value I want it modified to for this demonstration:
cursor.execute("""
UPDATE MetalPrices
SET Price = 7400
WHERE Metal = 'Copper' AND Date = '2010-01-01';
""")
```

To verify that the above Update has taken place, I retrieve the values from the MetalPrices Table:

```python
df = pd.read_sql("SELECT * FROM MetalPrices", connection)
df
```

Hence, the Update of the value has taken place, as reflected in the above DataFrame df.

Delete:

Making Deletion in the MetalPrices Table:

```python
# Deleting the data point at the locations that I want to delete for this demonstration through the filter of the 
# specified Metal and Date using the "WHERE" clause:
cursor.execute("""
DELETE FROM MetalPrices
WHERE Metal = 'Aluminum' AND Date = '2010-01-01';
""")
```

To verify that the above Delete has taken place, I retrieve the values from the MetalPrices Table:

```python
df = pd.read_sql("SELECT * FROM MetalPrices", connection)
df
```

Hence, the Deletion of the value has taken place, as reflected in the above DataFrame df.

In order to be sure that the effects of the above operations of the database are permanently saved to the database, the following commit() is called:

```python
# To permanently save the changes made to the database, the commit() is called:
connection.commit()
```

After running all the cells in this Notebook, if the database named metals is accessed through any other Notebook, all the changes made to the database from this Notebook here will also now reflect. Calling commit() above assures so.
